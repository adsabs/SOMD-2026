```python
# Statistics on the mentions and the relations

def get_all_entity_categories(data):
    
    """ 
    Get the list of named entity categories in a given dataset
    Args: input dataset (can be the training or testing set)
    """
    
    entity_categories = set(e["type"] for e in data)
    return entity_categories

def get_all_semantic_relations(data):

    """ 
    Get the list of all the semantic relations in a given dataset
    Args: input dataset (can be the training or testing set)
    """

    semantic_relations = []

    for e in data:
        relations = e["relations"]
        if relations != []:
            for r in relations:
                semantic_relations.append(r['type'])
    return set(semantic_relations)

def count_entity_distribution(data):

    """
    Distribution of each named entity category
    
    """

    entity_categories = get_all_entity_categories(data)
    count_labels = {key: 0 for key in entity_categories}

    for e in data:
        count_labels[e["type"]] += 1

    return dict(sorted(count_labels.items(),key=lambda item:item[1], reverse = True))
```

```python
def plot_entity_distribution(count_labels, figsize=(16, 8), save_path=None):
    """
    Plot entity distribution across subtasks (combined train+test counts).
    
    Parameters:
    -----------
    count_labels : dict
        Dictionary containing entity counts for each subtask and split
    figsize : tuple
        Figure size (width, height)
    save_path : str, optional
        Path to save the figure (e.g., 'entity_distribution.pdf')
    """
    # Set publication-quality style
    plt.rcParams['font.family'] = 'serif'
    plt.rcParams['font.size'] = 10
    plt.rcParams['axes.labelsize'] = 12
    plt.rcParams['axes.titlesize'] = 13
    plt.rcParams['xtick.labelsize'] = 9
    plt.rcParams['ytick.labelsize'] = 10
    plt.rcParams['legend.fontsize'] = 10
    plt.rcParams['legend.title_fontsize'] = 11
    
    # Extract all entity categories (assuming all subtasks have same entities)
    first_subtask = list(count_labels.keys())[0]
    entity_categories = list(count_labels[first_subtask].keys())
    
    # Filter out entities with zero counts across all subtasks and splits
    entities_with_data = []
    for entity in entity_categories:
        has_data = False
        for subtask in count_labels.keys():
            total = count_labels[subtask][entity]['train'] + count_labels[subtask][entity]['test']
            if total > 0:
                has_data = True
                break
        if has_data:
            entities_with_data.append(entity)
    
    n_subtasks = len(count_labels)
    n_entities = len(entities_with_data)
    
    # Create single plot
    fig, ax = plt.subplots(1, 1, figsize=figsize)
    
    # Color scheme for subtasks
    colors = ['#2E86AB', '#A23B72', '#F18F01']
    
    # Width of bars
    bar_width = 0.25
    
    # X positions
    x = np.arange(n_entities)
    
    # Plot bars for each subtask
    for idx, (subtask_name, subtask_data) in enumerate(count_labels.items()):
        # Sum train and test counts
        total_counts = [subtask_data[entity]['train'] + subtask_data[entity]['test'] 
                       for entity in entities_with_data]
        
        # Create bars with offset
        offset = (idx - (n_subtasks - 1) / 2) * bar_width
        bars = ax.bar(x + offset, total_counts, bar_width, 
                     label=subtask_name.replace('_', ' ').title(), 
                     color=colors[idx % len(colors)], alpha=0.85)
    
    # Customize plot
    ax.set_xlabel('Entity Type', fontweight='bold')
    ax.set_ylabel('Count', fontweight='bold')
    ax.set_title('Entity Distribution Across Subtasks', fontweight='bold', pad=15)
    ax.set_xticks(x)
    
    # Format x-axis labels
    entity_labels = [e.replace('_', '\n') for e in entities_with_data]
    ax.set_xticklabels(entity_labels, rotation=45, ha='right')
    
    # Add grid for easier reading
    ax.grid(axis='y', alpha=0.3, linestyle='--', linewidth=0.5)
    ax.set_axisbelow(True)
    
    # Add legend
    ax.legend(loc='upper right', frameon=True, fancybox=True, shadow=True, title='Subtask')
    
    # Use log scale if the range is very large
    max_count = max([count_labels[st][e]['train'] + count_labels[st][e]['test'] 
                     for st in count_labels.keys() for e in entities_with_data])
    if max_count > 1000:
        ax.set_yscale('log')
        ax.set_ylabel('Count (log scale)', fontweight='bold')
    
    plt.tight_layout()
    
    # Save if path is provided
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight', format=save_path.split('.')[-1])
        print(f"Figure saved to {save_path}")
    
    return fig
```

```python
# Are there any ambiguous mention (i.e., the same mention appearing in different clusters)?
```

```python
all_mentions = []
for l in train_labels_2:
    for mention_id in l:
        all_mentions.append(mention_id)
```

```python
mention_count = {key:0 for key in all_mentions}
```

```python
for mention in all_mentions:
    for cluster in train_labels_2:
        if mention in cluster:
            mention_count[mention] += 1
```

```python
for m, v in mention_count.items():
    if v >= 2:
        print(m)
```

```python
# Are there any mention type ambiguity (i.e., the same mention having different semantic type)?
```

```python
mention_types = {key:[] for key in all_mentions}
for mention in all_mentions:
    for d in train_data_2:
        if d["mention_id"] == mention:
            mention_type = d["type"]
            mention_types[mention].append(mention_type)
```

```python
for m, types in mention_types.items():
    if len(types) >= 2:
        print(m)
```
