```python
import json
from tqdm.auto import tqdm
```

```python
# load subtask 1 dataset
```

```python
subtask_1_train_data = []
with open('./data/SOMD 2026/subtask 1/train_data.jsonl', 'r') as f:
    for line in tqdm(f):
        subtask_1_train_data.append(json.loads(line))
```

```python
subtask_1_train_data[2]
```

```python
# for NER subsubtask what are the list of all types and relations
```

```python
subtask_1_types = set()
subtask_1_relations = set()

for e in subtask_1_train_data:
    subtask_1_types.update({e['type']})
    for r in e['relations']:
        subtask_1_relations.update({r['type']})
```
